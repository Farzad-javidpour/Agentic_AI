```python
from Utils.LLMMiddleware import OllamaLLM

code = "print((lambda f: f(f, 0))(lambda f, n: n if n > 10 else f(f, n + 1) + n))"
user_prompt = code
system_prompt = "You are a senior python developer and a good teacher. explain the input code for me in Farsi Language"

llm = OllamaLLM(model="gemma3:4b")

response = llm.chat(
    prompt= user_prompt,
    system_prompt= system_prompt
)

print(response)
```
