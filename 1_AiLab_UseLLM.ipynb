```python
from Utils.LLM_Skill import Skill
from Utils.LLM_ContextManager import ContextManager
from Utils.LLM_Client import OllamaClient

import json
from rich.markdown import Markdown
```

```python
good_assistant_skill = Skill(
    "skill",
    """You are a helpful assistant. 
    Give useful, short and courteus answers in JUST Farsi Language, no more extra explaine. Always be accurate. If you don't know the answer, say so."""
)
print(good_assistant_skill.get_skill())
```

```python
context = ContextManager()
context.add(
    "system",
    good_assistant_skill.get_skill()
)

context.add(
    "user",
    "یک فکت خنده دار تعریف کن"
)

print(context.get_context())
```

```python
client = OllamaClient()

result = client.chat(
    model="Gemma3:4b",
    messages= context.get_context()
)

print(json.dumps(result, indent=4, ensure_ascii=False)) 

Markdown(result["response"])
```
